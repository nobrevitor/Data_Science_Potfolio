# Detecção de Anomalias e Classificação de Fraudes em Transações Financeiras

### 📌 Objetivo do projeto
Fraude em transações financeiras continua sendo uma das ameaças mais graves do setor financeiro, causando muitas perdas anuais e prejudicando a confiança dos clientes.
Com o crescimento do comércio eletrônico e as transações online sendo a maior parte das transações financeiras nos dias atuais, detectar atividades fraudulentas de forma rápida e precisa é mais importante que nunca.

Este projeto tem como objetivo desenvolver um **modelo de machine learning** capaz de detectar transações fraudulentas utilizando o dataset **Bank Transaction Fraud Detection**.

Ao analizar padrões de transações e aplicar algoritmos avançados, nossos objetivos são:
- Aumentar a precisão na detecção de fraudes.
- Reduzir falsos positivos (transações legítmas identificadas incorretamente como fraude).
- Fornecer insights baseados em dados para aprimorar os sistemas de prevenção de fraudes.

---

### 📂 Descrição do Dataset

O conjunto de dados contém informações detalhadas sobre transações bancárias de clientes do LOL Bank Pvt. Ltd., com o objetivo de detectar atividades fraudulentas. Inclui detalhes demográficos dos clientes (por exemplo, ID do cliente, nome, gênero, idade, estado e cidade), especificações da transação (por exemplo, ID da transação, data, valor, tipo, comerciante, dispositivo e localização), bem como informações da conta e do banco (por exemplo, saldo da conta, tipo de conta, agência). 

Além disso, indica se uma transação é fraudulenta ou não. Esse conjunto de dados é usado para construir e treinar modelos para identificar transações fraudulentas com base em padrões, garantindo práticas bancárias seguras.

Este conjunto de dados não é sintético, mas sim uma base de dados reais obtidades de múltiplas fontes internas combinadas em um conjunto de dados.

**Source:** [Kaggle – Bank Transaction Fraud Detection](https://www.kaggle.com/datasets/marusagar/bank-transaction-fraud-detection/data)

#### *Variáveis:*


| Variável                | Descrição                                                                        | Tipo    |
|-------------------------|----------------------------------------------------------------------------------|---------|
| Customer_ID             | Identificador único para cada cliente no sistema do banco.                       | object  |
| Customer_Name           | Nome do cliente que realizou a transação.                                        | object  |
| Gender                  | Gênero do cliente (ex.: Masculino, Feminino, Outro).                             | object  |
| Age                     | Idade do cliente no momento da transação.                                        | int64   |
| State                   | Estado em que o cliente reside.                                                  | object  |
| City                    | Cidade em que o cliente vive.                                                    | object  |
| Bank_Branch             | Agência bancária específica onde o cliente possui conta.                         | object  |
| Account_Type            | Tipo de conta do cliente (ex.: Poupança, Corrente).                              | object  |
| Transaction_ID          | Identificador único para cada transação.                                         | object  |
| Transaction_Date        | Data em que a transação ocorreu.                                                 | object  |
| Transaction_Time        | Horário exato em que a transação foi iniciada.                                   | object  |
| Transaction_Amount      | Valor monetário da transação.                                                    | float64 |
| Merchant_ID             | Identificador único do comerciante envolvido na transação.                       | object  |
| Transaction_Type        | Tipo/natureza da transação (ex.: Saque, Depósito, Transferência).                | object  |
| Merchant_Category       | Categoria do comerciante (ex.: Varejo, Online, Viagem).                          | object  |
| Account_Balance         | Saldo da conta do cliente após a transação.                                      | float64 |
| Transaction_Device      | Dispositivo usado pelo cliente para realizar a transação (ex.: Mobile, Desktop). | object  |
| Transaction_Location    | Localização geográfica da transação (ex.: latitude, longitude).                  | object  |
| Device_Type             | Tipo de dispositivo usado para a transação (ex.: Smartphone, Laptop).            | object  |
| Transaction_Currency    | Moeda usada na transação (ex.: USD, EUR).                                        | object  |
| Customer_Contact        | Número de contato do cliente.                                                    | object  |
| Transaction_Description | Descrição breve da transação (ex.: compra, transferência).                       | object  |
| Customer_Email          | E-mail associado à conta do cliente.                                             | object  |
| Is_Fraud (**Target**)   | Indicador binário (1 ou 0) informando se a transação é fraudulenta ou não.       | int64   |
