# Detecção de Fraudes em Transações Financeiras

### 📌 Objetivo do projeto
Fraude em transações financeiras continua sendo uma das ameaças mais graves do setor financeiro, causando muitas perdas anuais e prejudicando a confiança dos clientes.
Com o crescimento do comércio eletrônico e as transações online sendo a maior parte das transações financeiras nos dias atuais, detectar atividades fraudulentas de forma rápida e precisa é mais importante que nunca.

Este projeto tem como objetivo desenvolver um **modelo de machine learning** capaz de detectar transações fraudulentas utilizando o dataset **[Fraud Detection Transactions Dataset](https://www.kaggle.com/datasets/samayashar/fraud-detection-transactions-dataset/data)**.

Ao analizar padrões de transações e aplicar algoritmos avançados, nossos objetivos são:
- Aumentar a precisão na detecção de fraudes.
- Reduzir falsos positivos (transações legítmas identificadas incorretamente como fraudes).
- Fornecer insights baseados em dados para aprimorar os sistemas de prevenção de fraudes.

---

### 📂 Descrição do Dataset

O conjunto de dados contém informações detalhadas sobre transações financeiras, incluindo identificação da transação e do usuário, valores, tipos de transação, saldo da conta, dispositivo e localização, além de atributos de risco e histórico de fraudes. Ele também inclui variáveis temporais e agregadas, como média de transações nos últimos 7 dias e contagem diária de transações.

Além disso, indica se uma transação é fraudulenta ou não. Esse conjunto de dados é usado para construir e treinar modelos para identificar transações fraudulentas com base em padrões, garantindo práticas bancárias seguras.

Este conjunto de dados é sintético e de alta qualidade, ou seja, os dados sensíveis presentes não pertencem a pessoas reais, estando em conformidade com a LGPD.

**Source:** [Kaggle – Fraud Detection Transactions Dataset](https://www.kaggle.com/datasets/samayashar/fraud-detection-transactions-dataset/data)

#### **Avaliar a distribuição da variável-alvo para verificar se há desbalanceamento:**

Como observado anteriormente, a distribuição de fraudes em cada variável do DataFrame gira em torno de 30%, indicando que a variável alvo apresenta desbalanceamento. Utilizar uma variável desbalanceada no treinamento do modelo pode resultar em uma métrica de acurácia enganosa, pois o modelo tenderia a classificar corretamente a maioria das transações não fraudulentas, enquanto teria desempenho ruim na identificação das transações fraudulentas — que, na prática, são o foco do modelo. Portanto, é fundamental aplicar técnicas de balanceamento quando as classes da variável alvo estiverem desbalanceadas, garantindo métricas mais realistas e desempenho efetivo na detecção de fraudes.

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='Fraud_Label', data=df, hue='Fraud_Label', palette='Set2', legend=False)
plt.title('Distribuição das classes da variável Fraud_Label', fontsize=14)
plt.xlabel('Fraud_Label')
plt.ylabel('Contagem')
plt.grid(axis='y', linestyle='--', alpha=0.5)

total = len(df)
for p in plt.gca().patches:
    height = p.get_height()
    plt.gca().annotate(f'{height} ({height/total:.2%})', 
                       (p.get_x() + p.get_width() / 2., height), 
                       ha='center', va='bottom', fontsize=10)

plt.show()
```

Como esperado, a variável alvo `Fraud_Label` apresenta um pequeno desbalanceamento, com as transações genuínas representando 67,87% dos dados, enquanto as transações fraudulentas correspondem a 32,13%. O desequilibrio apresentado será tratado com técnica de oversampling utilizando o SMOTE.

#### **Avaliar o Weigth of Evidence (WOE):**

O *Weight of Evidence (WOE)* é uma métrica amplamente utilizada em modelos de risco de crédito e detecção de fraude, pois permite avaliar o poder discriminatório de cada categoria de uma variável em relação ao evento de interesse (fraude ou inadimplência).

A principal vantagem do *WOE* é que ele transforma variáveis categóricas (e numéricas, quando agrupadas em faixas) em valores contínuos, preservando a relação com o target e facilitando a interpretação. Essa transformação gera insights claros sobre quais categorias aumentam ou reduzem a probabilidade de ocorrência de fraude, além de contribuir para a estabilidade e performance de modelos preditivos.

Na prática, bons valores de *WOE* indicam o quanto uma categoria consegue separar eventos de não-eventos. As regras práticas para interpretação são:

- **Próximo de 0** → categoria sem poder discriminatório relevante.

- **|*WOE*| entre 0.1 e 0.3** → separação fraca, mas potencialmente útil.

- **|*WOE*| entre 0.3 e 0.6** → separação moderada, já traz boa informação.

- **|*WOE*| acima de 0.6** → separação forte, categoria altamente relevante.

- ***WOE* muito extremo (>1 ou <-1)** → pode indicar problema de baixa representatividade da categoria (amostra pequena ou categoria rara).

Portanto, analisar o *WOE* é essencial para identificar variáveis com maior poder de discriminação, apoiar decisões de seleção de variáveis e evitar ruídos provenientes de categorias pouco representativas.

#### **Avaliar o Information Value (IV):**

O *Information Value (IV)* é uma métrica fundamental para avaliar o poder preditivo das variáveis em modelos de risco e de detecção de fraude. Ele quantifica o quanto uma variável contribui para a diferenciação entre classes — por exemplo, transações legítimas e fraudulentas.

Na prática, variáveis com *IV* mais altos têm maior capacidade de separar comportamentos distintos entre clientes ou transações, ajudando o modelo a identificar padrões de fraude de forma mais robusta. Isso é especialmente importante em cenários de fraude, onde pequenas diferenças de comportamento podem indicar atividades suspeitas.

Além disso, o *IV* auxilia no processo de seleção de variáveis, permitindo eliminar variáveis irrelevantes ou redundantes que pouco agregam ao modelo. Isso melhora a eficiência computacional e reduz o risco de sobreajuste.

#### **Feature Engineering:**

Na etapa de *EDA*, ao analisar as métricas *WOE (Weight of Evidence*) e *IV (Information Value)*, observou-se que as variáveis `Failed_Transaction_Count_7d` e `Risk_Score` apresentaram valores extremamente elevados. 

Para preservar o potencial preditivo da variável `Failed_Transaction_Count_7d`, mas ao mesmo tempo mitigar riscos de data leakage ou de distorções no modelo, optou-se pela criação de novas variáveis derivadas, capazes de capturar diferentes aspectos da sua distribuição sem comprometer a informação relevante que carrega. 

A variável `Risk_Score` será excluída, pois pode ter sido derivada de outro modelo ou representar informações diretamente relacionadas à ocorrência de fraude. Isso faria com que o modelo treinado se tornasse dependente de informações externas ou excessivamente acurado por conter indícios do resultado futuro da transação, caracterizando *data leakage*.

As novas variáveis serão construídas da seguinte forma:

*Variável de tempo:*

- `Is_Night`: Variável binária (0 = dia, 1 = noite).

*Histórico por usuário:*

- `Transaction_Amount_mean`: Média do valor das transações por usuário.

- `Transaction_Amount_std`: Desvio padrão do valor das transações por usuário.

- `Transaction_Amount_max`: Valor máximo de transação por usuário.

- `Transaction_Amount_min`: Valor mínimo de transação por usuário.

- `Failed_Transaction_Count_7d_mean`: Média de transações falhas nos últimos 7 dias por usuário.

- `Failed_Transaction_Count_7d_max`: Máximo de transações falhas nos últimos 7 dias por usuário.

*Variáveis de taxa (incluindo `Failed_Transaction_Count_7d`):*

- `Failed_Transaction_Rate`: Proporção de transações falhas por dia.

- `Transaction_Ratio`: Proporção de transações realizadas por dia.

```python
df_feature = df_treat.copy()

df_feature['Timestamp'] = pd.to_datetime(df_feature['Timestamp'])
df_feature = df_feature.sort_values(by=['User_ID', 'Timestamp'])

# Features temporais
df_feature['Is_Night'] = df_feature['Timestamp'].dt.hour.between(0, 6).astype(int)

# Agregações por usuário (histórico)
agg_funcs = {
    'Transaction_Amount': ['mean', 'std', 'max', 'min'],
    'Failed_Transaction_Count_7d': ['mean', 'max']
}
user_agg = df_feature.groupby('User_ID').agg(agg_funcs)
user_agg.columns = ['_'.join(col) for col in user_agg.columns]
user_agg.reset_index(inplace=True)
user_agg.fillna({'Transaction_Amount_std': 0}, inplace=True)
df_feature = df_feature.merge(user_agg, on='User_ID', how='left')

# Features de taxa
df_feature['Failed_Transaction_Rate'] = df_feature['Failed_Transaction_Count_7d'] / (df_feature['Daily_Transaction_Count'] + 1)
df_feature['Transaction_Ratio'] = df_feature['Daily_Transaction_Count'] / (df_feature['Avg_Transaction_Amount_7d'] + 1)

df_feature.head()
```

#### **Seleção de variáveis com Feature Importance:**

Para otimizar o desempenho do modelo e reduzir ruídos, serão selecionadas as variáveis de maior relevância. Para isso, será criado um modelo específico de seleção de variáveis, utilizando o atributo `feature_importances_` do próprio *XGBClassifier* como critério de importância.

```python
smote = SMOTE(random_state=23)
X_train_res, y_train_res = smote.fit_resample(X_train, y_train)

model_select = XGBClassifier(
    objective='binary:logistic',
    eval_metric='logloss',
    random_state=23
)

model_select.fit(X_train_res, y_train_res)

feat_importances = pd.DataFrame({
    'feature': X_train.columns,
    'importance': model_select.feature_importances_
}).sort_values(by='importance', ascending=False)

plt.figure(figsize=(12,8))
sns.barplot(data=feat_importances.head(20), x='importance', y='feature')
plt.title('Features por Importância (XGBoost)')
plt.xlabel('Importância')
plt.ylabel('Features')
plt.grid(True, axis='x', linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()

top_features = feat_importances[feat_importances['importance'] > 0.05]['feature'].tolist()
```

Para validar as variáveis selecionadas com `feature_importances_` — que avalia o quanto cada variável contribui para a melhoria da função de custo em divisões da árvore — será utilizado o *SHAP (Shapley Additive exPlanations)*. O *SHAP* permite medir a contribuição marginal e individual de cada variável em cada predição, tornando a seleção de variáveis mais robusta e confiável. Podendo assim ser generalizada para outros algorítmos.

```python
explainer = shap.TreeExplainer(model_select)
shap_values = explainer.shap_values(X_test)

shap.summary_plot(shap_values, X_test)
```

A seleção das variáveis para a modelagem será realizada a partir da interseção entre aquelas identificadas pelo `feature_importances_` com importância superior a 0.05 e as 12 mais relevantes segundo o *SHAP*. Dessa forma, garante-se que apenas as variáveis com impacto consistente em diferentes métodos sejam utilizadas no modelo.

```python
shap_importance = np.abs(shap_values).mean(axis=0)

shap_importance_df = pd.DataFrame({
    'feature': X_train.columns,
    'importance': shap_importance
}).sort_values('importance', ascending=False)

top_features_shap = shap_importance_df.head(12)['feature'].tolist()
intersecao = [item for item in top_features if item in top_features_shap]

print(f'A interseção contém {len(intersecao)} variáveis.\n\nElas são:\n')    
for var in intersecao:
    print(var)    
```

#### **Random Forest:**

Como o problema é de classificação, no Random Forest será adotado o critério `entropy`, uma vez que ele se mostra mais adequado em cenários que exigem uma separação precisa entre classes em regiões incertas ou desbalanceadas, por ser mais rigoroso na avaliação da impureza. O desempenho do modelo será avaliado por meio das métricas *Precision*, *Recall*, *F1-score*, *Accuracy*, *ROC-AUC* e *Log Loss*.

```python
rf_model = RandomForestClassifier(
    criterion='entropy',
    random_state=23,
    n_jobs=-1
)

rf_model.fit(X_train_res, y_train_res)

y_pred_rf = rf_model.predict(X_test)
y_proba_rf = rf_model.predict_proba(X_test)[:, 1]

print('Classification Report:\n\n', classification_report(y_test, y_pred_rf))
print(f'ROC-AUC: {roc_auc_score(y_test, y_proba_rf):.4f}')
print(f'Log Loss: {log_loss(y_test, y_proba_rf):.4f}')

cm = confusion_matrix(y_test, y_pred_rf)

plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=['Não Fraude (0)', 'Fraude (1)'],
            yticklabels=['Não Fraude (0)', 'Fraude (1)'])

plt.xlabel('Previsto')
plt.ylabel('Real')
plt.title('Matriz de Confusão')
plt.show()
```

Também, a taxa de falsos positivos ainda é elevada, o que pode comprometer a experiência do cliente, gerar perda de receita e aumentar os custos operacionais, uma vez que transações legítimas classificadas incorretamente como fraude exigem revisão manual.

#### **Gradient Boosting (XGBoost):**

Para otimizar o desempenho do *Gradient Boosting* no problema de classificação, será utilizado o hiperparâmetro `objective`='binary:logistic', definindo que o modelo deve tratar a tarefa como classificação binária. O parâmetro `eval_metric`='logloss' será utilizado para avaliar o desempenho do modelo com base no *Log Loss*, permitindo melhorar a qualidade das previsões probabilísticas.

#### **GridSearchCV para encontrar os melhores hiperparametros:**

Com o algoritmo selecionado, é necessário ajustar alguns hiperparâmetros para melhorar o desempenho do modelo, sem depender apenas das configurações automáticas. Para não aumentar excessivamente o custo computacional, serão avaliados apenas os seguintes valores para cada hiperparâmetro:

**n_estimators:**
- Define o número de árvores que o modelo vai construir.
- Valores avaliados: [100, 200, 500]

**max_depth:** 
- Determina a profundidade máxima de cada árvore.
- Valores avaliados: [3, 5]

**learning_rate:**
- Taxa de aprendizado que controla o quanto cada árvore contribui para a previsão final.
- Valores avaliados: [0.05, 0.1, 0.2]

**subsample:**
- Proporção de amostras utilizadas para treinar cada árvore.
- Valores avaliados: [0.8, 1.0]

**colsample_bytree:**
- Proporção de colunas (features) utilizadas para construir cada árvore.
- Valores avaliados: [0.8, 1.0]

**min_child_weight:**
- Soma mínima dos pesos (ou instâncias) necessárias em uma folha da árvore.
- Valores avaliados: [1, 3]

Para essa seleção, será utilizado o *GridSearchCV* com o scoring baseado no *F1-score*, já que essa métrica equilibra *Precision* e *Recall*. Dessa forma, é possível encontrar uma combinação de hiperparâmetros que melhore o desempenho geral do modelo, sem aumentar excessivamente a quantidade de falsos positivos ou falsos negativos.

```python
param_grid = {
    'n_estimators': [100, 200, 500],
    'max_depth': [3, 5],
    'learning_rate': [0.05, 0.1, 0.2],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0],
    'min_child_weight': [1, 3]
}

xgb = XGBClassifier(
    objective='binary:logistic',
    eval_metric='logloss',
    random_state=23
)

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=23)

grid = GridSearchCV(
    estimator=xgb,
    param_grid=param_grid,
    scoring='f1',
    cv=skf,
    verbose=1,
    n_jobs=-1
)
grid.fit(X_train_res, y_train_res)

print(f'\nMelhores parâmetros: {grid.best_params_}\n')
print(f'Melhor F1-score: {grid.best_score_}')

best_xgb = grid.best_estimator_
best_xgb.fit(X_train_res, y_train_res)
```

---

#### **Melhor Threshold que aumente o *F1-score*:**

Após encontrar a melhor combinação de hiperparâmetros, uma etapa importante é a definição do *threshold*, que determina o limite de probabilidade para classificar uma transação como fraude ou não. Ajustar esse valor permite reduzir a quantidade de falsos negativos, tornando o modelo mais robusto, pois menos fraudes serão classificadas incorretamente como não fraudulentas.

O *threshold* será definido pelo valor que melhore a métrica *F1-score*.

```python
y_probs = best_xgb.predict_proba(X_train_res)[:, 1]
    
precision, recall, thresholds = precision_recall_curve(y_train_res, y_probs)

f1_scores = 2 * (precision[:-1] * recall[:-1]) / (precision[:-1] + recall[:-1] + 1e-9)

best_idx = np.argmax(f1_scores)
best_threshold = thresholds[best_idx]
best_f1 = f1_scores[best_idx]

print(f'Melhor Threshold: {best_threshold:.3f}')
print(f'F1-score correspondente: {best_f1:.3f}')
```

O *threshold* ótimo encontrado foi 0,46, valor que maximiza o *F1-score* e reduz a ocorrência de falsos negativos. Assim, toda transação com probabilidade prevista acima de 0,46 é classificada como fraude.

##### **Variáveis categóricas:**

Ao analisar as variáveis categóricas utilizadas na modelagem, observamos que algumas categorias apresentam maior probabilidade de estarem associadas a transações fraudulentas em comparação com outras. Esse padrão permite identificar perfis de usuários mais suscetíveis a fraudes. Entre as variáveis categóricas avaliadas, destacam-se as seguintes categorias como as mais vulneráveis:
<br><br>

**Variável `Transaction_Type`:** 

`ATM Withdrawal`

Quanto ao tipo de transação, os caixas eletrônicos (ATMs) representam um ponto crítico para golpes e fraudes, especialmente durante saques em dinheiro. Por estarem expostos fora das agências bancárias, tornam-se alvos frequentes de criminosos. Um exemplo é o uso de dispositivos conhecidos como “chupa-cabras”, instalados no leitor do cartão para capturar dados da tarja magnética ou do chip. Além disso, câmeras ocultas podem ser posicionadas estrategicamente para registrar a digitação do PIN, comprometendo a segurança do usuário.
<br><br>

**Variável `Device_Type`:** 

`Tablet`

O celular conta com uma arquitetura de segurança mais robusta: nos sistemas móveis (iOS e Android), os aplicativos rodam em ambientes isolados, o que impede que um app malicioso acesse dados de outros. Além disso, modelos mais recentes oferecem autenticação biométrica, como impressão digital ou reconhecimento facial, adicionando uma camada extra de proteção.

O tablet se assemelha ao celular em termos de segurança, já que geralmente também utiliza sistemas móveis com isolamento de aplicativos. No entanto, por ser um dispositivo maior e, muitas vezes, compartilhado, há maior risco de uso por terceiros, diferentemente do celular, que costuma estar sempre com o dono.

Já o notebook apresenta mais vulnerabilidades, pois sua arquitetura permite a instalação de programas de diversas origens, ampliando a superfície de ataque para vírus e malwares. Além disso, muitos modelos ainda não contam com autenticação biométrica, o que reduz seu nível de segurança em comparação aos dispositivos móveis.

Por fim, é importante ressaltar que, independentemente do aparelho utilizado, a segurança das transações financeiras depende fortemente do comportamento do usuário. Golpes como *phishing* exploram principalmente a falta de atenção e o descuido durante compras online, tornando o fator humano o maior ponto de risco.
<br><br>

**Variável `Authentication_Method`:**

`OTP`

A autenticação OTP (One-Time Password) é um método baseado em senhas de uso único, que gera um código temporário e exclusivo para verificar a identidade do usuário. Esse código é válido apenas para uma sessão de login ou transação e expira automaticamente após o uso ou em poucos minutos.

Esse recurso costuma ser utilizado em conjunto com PINs ou senhas fixas, oferecendo maior segurança do que depender apenas desses mecanismos. No entanto, ainda não supera a confiabilidade da autenticação biométrica, considerada o método mais seguro atualmente.

Entre as vantagens do OTP está a adição de uma camada extra de proteção, reduzindo a dependência exclusiva da senha estática do usuário. Porém, sua segurança depende diretamente do canal de entrega, ficando suscetível a riscos como:

**Vulnerabilidade na entrega:** mensagens enviadas por SMS, e-mail, WhatsApp ou ligação podem ser interceptadas, atrasadas ou perdidas por falhas de rede.

**Ataques de engenharia social:** criminosos podem convencer o usuário a revelar o código OTP por meio de phishing (sites falsos, e-mails fraudulentos, perfis falsos de empresas) ou até por ligações que exploram o medo e a urgência, levando a vítima a fornecer o código sem refletir. Uma vez de posse desse dado, o invasor tem acesso imediato à conta.

**Dependência de dispositivos externos:** é necessário ter sempre disponível o dispositivo que recebe o OTP (como smartphone ou token físico). A perda, roubo ou dano desse dispositivo pode impedir temporariamente o acesso à conta, exigindo procedimentos de recuperação.

Em resumo, embora o OTP seja mais seguro que métodos baseados apenas em senhas, sua principal vulnerabilidade ainda está no comportamento do usuário. A autenticação fornece uma camada extra de proteção, mas sua eficácia depende diretamente da atenção e dos cuidados do próprio usuário ao lidar com tentativas de fraude.
<br><br>

**Variável `Merchant_Category`:** 

`Restaurants`

Restaurantes representam um vetor de risco quando a segurança da rede à qual as máquinas de cartão estão conectadas é frágil. Redes mal configuradas — com portas desnecessárias abertas, falta de segmentação entre dispositivos e ausência de controles de acesso — podem ser facilmente comprometidas. Se o Wi-Fi aberto ao cliente compartilhar a mesma rede que os terminais POS, um atacante pode obter acesso ao tráfego e aos registros de transações; nisso incluem-se tentativas de interceptação e clonagem de cartões. Por isso, práticas como segmentação de redes, autenticação forte, criptografia de tráfego e conformidade com padrões (ex.: PCI-DSS) são essenciais para reduzir esse risco.
<br><br>

**Variável `Card_Type`:** 

`Discover`

Esse atraso na massificação dessas tecnologias — sobretudo em mercados internacionais — abriu brechas para ataques em regiões onde o chip ou métodos de autenticação mais fortes ainda não estavam disponíveis.

##### **Boas práticas de segurança para os clientes:**

**Uso de caixas eletrônicos (ATMs):**

- Verifique se o equipamento apresenta sinais de adulteração, especialmente no local de inserção do cartão, e fique atento à presença de câmeras suspeitas.

- Observe o ambiente ao redor para evitar aproximação de pessoas tentando obter informações confidenciais.

- Nunca aceite ajuda de desconhecidos para realizar transações.

**Preferência por dispositivos seguros:**

- Dê preferência a dispositivos móveis de uso pessoal, que geralmente oferecem maior proteção.

- Evite clicar em links suspeitos ou acessar sites maliciosos que possam estar associados a ataques de phishing.

**Métodos de autenticação:**

- Sempre que possível, utilize biometria ou autenticações em múltiplos fatores, em vez de apenas senhas estáticas.

**Locais de compra e conexão:**

- Evite realizar transações financeiras em redes Wi-Fi públicas, onde a segurança não é garantida.
